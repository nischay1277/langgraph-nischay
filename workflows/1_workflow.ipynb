```python
llm = ChatOllama(model="llama3.2", temperature=0.5)
```

```python
hf_token=os.getenv("HF_TOKEN")
```

```python
class LLMstate(TypedDict):
    question:str
    answer:str
```

```python
def findc(state:LLMstate)->LLMstate:
    question=state['question']
    res=llm.invoke(question)
    state['answer']=res.content
    return state

```

```python
def findc2(state:LLMstate)->LLMstate:
    question=state['answer']
    res=llm.invoke(question)
    state['question']=res.content
    return state
```

```python
graph=StateGraph(LLMstate)
graph.add_node('LLMstate',findc)
graph.add_node('stage2',findc)
graph.add_edge(START,'LLMstate')
graph.add_edge('LLMstate','stage2')
graph.add_edge('stage2',END)
workflow=graph.compile()
result=workflow.invoke({'question':'what is spiderman and is it correct?','answer':''})
print(result)

```

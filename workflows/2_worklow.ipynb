```python
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
from langchain_ollama import ChatOllama

```

```python
class bat(TypedDict):
    runs:int
    fours:int
    six:int
    balls:int

    sr:float
    bpb:float
    bp:float
    summar:str


```

```python
def calcsr(state:bat)->bat:
    state['sr']=(state['runs']/state['balls'])/100
    return {'sr':'sr'}

```

```python
def calcbp(state:bat)->bat:
    state['bpb']=state['balls']/(state['fours']+state['six'])
    return {'bpb':bpb}
```

```python
def calcbpb(state:bat)->bat:
    state['bp'] = (
        ((state['fours'] * 4) + (state['six'] * 6))
        / state['runs']
    ) * 100
    return {'bp':bp}

```

```python
def summ(state:bat)->bat:
    summary=f"""strike rate is {state['sr']}\n bp is {state['bp']}\n bpb is {state['bpb']}"""
    state['summar']=summary
    return {'summar':'summary'}
```

```python
graph=StateGraph(bat)
graph.add_node('strike_rate',calcsr)
graph.add_node('bpp',calcbp)
graph.add_node('bpbb',calcbpb)
graph.add_node('summary',summ)
graph.add_edge(START,'strike_rate')
graph.add_edge('strike_rate','summary')
graph.add_edge(START,'bpp')
graph.add_edge(START,'bpbb')
graph.add_edge('bpp','summary')
graph.add_edge('bpbb','summary')
graph.add_edge('summary',END)
workflow=graph.compile()
initial={'runs':100,'balls':50,'fours':6,'six':4}
result=workflow.invoke(initial)
print(result)



```

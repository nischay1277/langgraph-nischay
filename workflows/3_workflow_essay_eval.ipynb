```python
from langgraph.graph import StateGraph ,START,END  
from langchain_ollama import ChatOllama
from pydantic import BaseModel,Field


```

```python
class LLMschema(BaseModel):
    feedback:str =Field(description='detailed feedback for the essay')
    score :int =Field(description="score out of 10 for the feedback",ge=0 ,le=10)
    
```

```python
essay="""Education Is the Most Powerful Weapon to Change the World
Education is the foundation of individual growth and national development. It empowers people with knowledge, develops critical thinking, and enables them to make informed decisions. As Nelson Mandela famously emphasised, education can transform both individuals and societies.
In India, education has played a significant role in reducing poverty, promoting gender equality, and creating opportunities for social mobility. Government initiatives such as the Right to Education Act seek to make education more accessible to children. However, challenges such as unequal access, inadequate infrastructure, and the digital divide continue to affect educational outcomes.
Education must therefore go beyond memorising facts and focus on creativity, practical skills, ethical values, and scientific temper. Teachers, governments, and communities must work together to ensure that quality education reaches every section of society.
Conclusion: Education is not merely a means of securing employment; it is an instrument of empowerment and social transformation. A nation that invests in education invests in its future, because educated citizens form the foundation of a just, progressive, and prosperous society."""
prompt=f'evaluate this essay from 1-10. give an honest rating {essay}'
```

```python
print("Python is working")
```

```python
model=ChatOllama(
    model="llama3.2",
    temperature=0
)
structured_model=model.with_structured_output(LLMschema)
result=structured_model.invoke(prompt)
print(result)
```
